```python
def extract_and_save_chromosome_column(adata_file_path, output_file_path, column_name):
    adata = sc.read_h5ad(adata_file_path)
    counts = adata.X
    counts = pd.DataFrame(counts, index=adata.obs['Unique_ID'], columns=adata.var.index)
    extracted_data = counts[[column_name]]
    print(extracted_data.shape)
    extracted_data.to_csv(output_file_path, index=True)
```

```python
extract_and_save_chromosome_column(
    adata_file_path=f"{indir}/5hmC_chrom100k.CG.pass_paired_QC.h5ad",
    output_file_path=f"{outdir}/11-extracted_chr5_201_5hmCG.csv",
    column_name="chr5_201"
)

extract_and_save_chromosome_column(
    adata_file_path=f"{indir}/5hmC_chrom100k.CG.pass_paired_QC.h5ad",
    output_file_path=f"{outdir}/11-extracted_chr2_390_5hmCG.csv",
    column_name="chr2_390"
)

extract_and_save_chromosome_column(
    adata_file_path=f"{indir}/5hmC_chrom100k.CG.pass_paired_QC.h5ad",
    output_file_path=f"{outdir}/11-extracted_chr8_173_5hmCG.csv",
    column_name="chr8_173"
)

extract_and_save_chromosome_column(
    adata_file_path=f"{indir}/5hmC_chrom100k.CG.pass_paired_QC.h5ad",
    output_file_path=f"{outdir}/11-extracted_chr8_171_5hmCG.csv",
    column_name="chr8_171"
)

extract_and_save_chromosome_column(
    adata_file_path=f"{indir}/5hmC_chrom100k.CG.pass_paired_QC.h5ad",
    output_file_path=f"{outdir}/11-extracted_chr8_172_5hmCG.csv",
    column_name="chr8_172"
)
```
