```r
TSO_metadata <- read.csv("../../input/01-youth/RNA_DNA_match_name_QC_class_label.csv", row.names = 1)
```

```r
TSO_QC_metadata <- subset(TSO_metadata ,total_QC == 1 )
```

```r
TSO_QC_metadata
```

```r
dim(TSO_QC_metadata)
```

```r
TSO_QC_df=TSO_QC_metadata[c("hmC","mC","subclass_label")]
```

```r
allc_hmC <- read.table("../../input/01-youth/TSO-joint-RNA_5hmC_allc_table_0820_47712cell.txt")
colnames(allc_hmC) <- c("hmC","AllcPath_hmC")
allc_hmC$hmC <- gsub("^allc_","",allc_hmC$hmC)
```

```r
allc_mC <- read.table("../../input/01-youth/TSO-joint-RNA_5mC_allc_table_0820_47712cell.txt")
colnames(allc_mC) <- c("mC","AllcPath_mC")
allc_mC$mC <- gsub("^allc_","",allc_mC$mC)
```

```r
TSO_QC_df<-merge(TSO_QC_df,allc_hmC,by = "hmC",all=FALSE)
TSO_QC_df<-merge(TSO_QC_df,allc_mC,by = "mC",all=FALSE)
```

```r
TSO_QC_df
```

```r
TSO_QC_df_split=split(TSO_QC_df, TSO_QC_df$subclass_label)
```

```r
names(TSO_QC_df_split)
```

```r
# iterate over all subclass
for (i in names(TSO_QC_df_split)) {
  # get the allc_path for the current subclass
  allc_paths <- TSO_QC_df_split[[i]]$AllcPath_hmC
  # generate the filename for saving
  file_name <- paste0(indir,"/5hmC_",gsub("[ /]", ".", i), ".txt")
  # do the saving
  write.table(allc_paths, file = file_name, row.names = FALSE, col.names = FALSE,quote = FALSE)
  cat("Saved", file_name, "\n")
    
  # the same procedure for 5mC instead of 5hmC
  allc_paths <- TSO_QC_df_split[[i]]$AllcPath_mC
  file_name <- paste0(indir,"/5mC_",gsub("[ /]", ".", i), ".txt")
  write.table(allc_paths, file = file_name, row.names = FALSE, col.names = FALSE,quote = FALSE)
  cat("Saved", file_name, "\n")
}
```
