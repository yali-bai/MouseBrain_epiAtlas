```python
cor_df=pd.read_csv("../../input/01-youth/correspondence_of_subclass_and_class_for_integration_with_zeng.csv")
```

```python
cor_df
```

```python
def get_threeclass(class_name):
    class_suffix=class_name.split()[-1]
    if class_suffix in ["GABA", "Glut"]:
        return class_suffix
    else:
        return "non"
```

```python
cor_df["threeclass"]=cor_df["class"].apply(get_threeclass)
```
