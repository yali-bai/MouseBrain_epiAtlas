```r
Joint_Cabernet_RNA_subclass_mean <- readRDS(paste0(indir,"/02-Joint_Cabernet_subclass_mean_dat_final.rds"))
```

```r
Joint_Cabernet_gene_variances_all <- apply(Joint_Cabernet_RNA_subclass_mean, 1, var)
zeng_gene_variances_all <- apply(zeng_RNA_subclass_mean, 1, var)
```

```r
head(Joint_Cabernet_gene_variances_all);length(Joint_Cabernet_gene_variances_all)
head(zeng_gene_variances_all);length(zeng_gene_variances_all)
```

```r
Joint_Cabernet_gene_variances_df_all <- data.frame(Gene = rownames(Joint_Cabernet_RNA_subclass_mean), Joint_Cabernet_variance = Joint_Cabernet_gene_variances_all)
head(Joint_Cabernet_gene_variances_df_all);dim(Joint_Cabernet_gene_variances_df_all)

zeng_gene_variances_df_all <- data.frame(Gene = rownames(zeng_RNA_subclass_mean), zeng_variance = zeng_gene_variances_all)
head(zeng_gene_variances_df_all);dim(zeng_gene_variances_df_all)
```

```r
merged_gene_variances_df_all <- merge(Joint_Cabernet_gene_variances_df_all, zeng_gene_variances_df_all, by = "Gene")
head(merged_gene_variances_df_all);dim(merged_gene_variances_df_all)
```

```r
write.csv(merged_gene_variances_df_all,paste0(outdir,"/22-Joint_Cabernet_zeng_Gene_Variance_subclass.csv"),quote=F,row.names=F)
```

```r
metaData <- read.csv("../../../04.data/05.intermediate_files/01.RNA/01.Young_Mouse/all_gene_upset_group_list_new.csv")
head(metaData);dim(metaData)
```

```r
library(dplyr)
```

```r
metaData <- metaData %>% dplyr::filter(mc_type %in% "CG",var_dim %in% "genebody")
```

```r
head(metaData);dim(metaData)
```

```r
Joint_Cabernet_RNA_subclass_mean_filter <- Joint_Cabernet_RNA_subclass_mean[metaData$gene_id,]
head(Joint_Cabernet_RNA_subclass_mean_filter);dim(Joint_Cabernet_RNA_subclass_mean_filter)
```

```r
zeng_RNA_subclass_mean_filter <- zeng_RNA_subclass_mean[metaData$gene_id,]
head(zeng_RNA_subclass_mean_filter);dim(zeng_RNA_subclass_mean_filter)
```

```r
Joint_Cabernet_gene_variances <- apply(Joint_Cabernet_RNA_subclass_mean_filter, 1, var)
zeng_gene_variances <- apply(zeng_RNA_subclass_mean_filter, 1, var)
```

```r
head(Joint_Cabernet_gene_variances);length(Joint_Cabernet_gene_variances)
head(zeng_gene_variances);length(zeng_gene_variances)
```

```r
Joint_Cabernet_gene_variances_df <- data.frame(Gene = rownames(Joint_Cabernet_RNA_subclass_mean_filter), Joint_Cabernet_variance = Joint_Cabernet_gene_variances)
head(Joint_Cabernet_gene_variances_df);dim(Joint_Cabernet_gene_variances_df)

zeng_gene_variances_df <- data.frame(Gene = rownames(zeng_RNA_subclass_mean_filter), zeng_ariance = zeng_gene_variances)
head(zeng_gene_variances_df);dim(zeng_gene_variances_df)
```

```r
merged_gene_variances_df <- merge(Joint_Cabernet_gene_variances_df, zeng_gene_variances_df, by = "Gene")
head(merged_gene_variances_df);dim(merged_gene_variances_df)
```

```r
merged_data <- merged_gene_variances_df %>%
  left_join(metaData, by = c("Gene" = "gene_id"))
head(merged_data);dim(merged_data)
```

```r
p01 <- ggplot(merged_data, aes(x = zeng_ariance, y = Correlation_5mC)) +
  geom_point(color = "black", alpha = 0.9, size = 1.2) +
  stat_density_2d(aes(fill = ..level..), geom = "polygon", alpha = 0.9) +
  scale_fill_gradientn(colors = c("black", "red", "yellow"), limits = c(0, 200), oob = scales::squish) +
  labs(x = "Variance", y = "Correlation (5mCG)", fill = "Density") +
  theme_minimal() +
  theme(
    legend.position = "right",
    panel.border = element_rect(color = "black", fill = NA, size = 1),
    axis.text.x = element_text(color = "black", size = 15),
    axis.text.y = element_text(color = "black", size = 15),
    axis.title.x = element_text(size = 16, face = "bold"),
    axis.title.y = element_text(size = 16, face = "bold"),
    legend.text = element_text(size = 15),
    legend.title = element_text(size = 16, face = "bold")
  )
```

```r
options(repr.plot.width = 6.45, repr.plot.height = 5, repr.plot.res = 100)
p01
```
