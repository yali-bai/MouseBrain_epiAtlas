```r
# "indir" is a custom input path, and "outdir" is a custom output path.
# indir=""
outdir="../../../03.data/02.metainfo/01.Young_Mouse/"
```

```r
regions <- c("CB","CTXsp","HPF","HY","Isocortex-1","Isocortex-2","MB","MY","OLF","PAL","P","STR","TH")
regions
```

```r
terms <- NULL
```

```r
for(i in regions){
  object <- read.delim(sprintf("%s/%s_meta.txt",indir,i),header = T) 
  if(is.null(terms)){
    terms <- data.frame(region=i, object)
  }else{
    terms <- rbind(terms, data.frame(region=i, object))
  }
}
```

```r
head(terms);dim(terms)
```

```r
reserved_region <- c("MO-FRP", "MOp", "SS-GU-VISC", "AUD-TEa-PERI-ECT", "VIS-PTLp", "VIS", "ACA", "PL-ILA-ORB", "AI", "RSP", "OLF", "HIP", "STRd", "STRv")
reserved_region
length(reserved_region)
```

```r
table(reserved_region %in% zeng_v3_metadata$region_label)
```

```r
zeng_v3_metadata_filter <- zeng_v3_metadata[zeng_v3_metadata$region_label %in% reserved_region,]
head(zeng_v3_metadata_filter);dim(zeng_v3_metadata_filter)
```

```r
count_subclass <- zeng_v3_metadata_filter %>%
  group_by(subclass_label) %>%
  summarise(count = n())
count_subclass
```

```r
count_subclass[count_subclass$count > 100,]
count_subclass[count_subclass$count > 100,]$subclass_label
count_subclass[count_subclass$count > 100,]$subclass_label -> subclass_100
```

```r
zeng_v3_metadata_filter2 <- zeng_v3_metadata_filter[zeng_v3_metadata_filter$subclass_label %in% subclass_100,]
head(zeng_v3_metadata_filter2);dim(zeng_v3_metadata_filter2)
```
