```r
library(Seurat)
```

```r
# "indir" is a custom input path, and "outdir" is a custom output path.
indir="../../../03.data/02.metainfo/01.Young_Mouse/"
# outdir=""
```

```r
regions <- c("CB","CTXsp","HPF","HY","Isocortex_1","Isocortex_2","MB","MY","OLF","PAL","PO","STR","TH")
regions
```

```r
all_cell_used <- readRDS(sprintf("%s/01-zeng_v3_metadata_downsample_1000.rds",indir)) 
head(all_cell_used);dim(all_cell_used)
```

```r
h5ad_process <- function(region){
    counts <- readRDS(sprintf("%s/%s.rds",indir,region)) 
    all_cell_used <- all_cell_used[all_cell_used$cell_new %in% colnames(counts),]
    selected_cells <- all_cell_used$cell_new
    counts_subset <- counts[, selected_cells]
    saveRDS(counts_subset,sprintf("%s/%s_counts.rds",outdir,region)) 
}
```

```r
for (i in regions){
    print(i)
    h5ad_process(i)
}
print("All regions finished")
```
