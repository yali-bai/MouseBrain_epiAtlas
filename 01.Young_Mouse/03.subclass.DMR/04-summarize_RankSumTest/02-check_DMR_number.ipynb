```python
tests=[]
for i in os.listdir("."):
    if os.path.isdir(i) and i[0] in "0":
        tests.append(i)
tests=sorted(tests)
tests
```

```python
modifications=["5mCG", "5hmCG"]
```

```python
states=["hyper", "hypo"]
```

```python
for test_ in tests:
    df_stat_list=[]
    for modification_ in modifications:
        df_states=ad.read_h5ad(f"{indir}/{test_}/{modification_}_DMR_states_{test_.split("_")[-2]}_1vsOthers.h5ad").to_df()
        number_stats_=pd.DataFrame(pd.DataFrame([
                        (df_states==1).sum(axis=1),
                        (df_states==-1).sum(axis=1),
                        (df_states.abs()==1).sum(axis=1)]).T)
        number_stats_.columns=[f"{modification_} hyper", f"{modification_} hypo", f"{modification_} total"]
        number_stats_=number_stats_._append(
            pd.Series([np.sum((df_states==1).sum(axis=0)>0),
                       np.sum((df_states==-1).sum(axis=0)>0),
                       np.sum((df_states.abs()==1).sum(axis=0)>0)],
                       index=[f"{modification_} hyper", f"{modification_} hypo", f"{modification_} total"],
                       name="total"))
        df_stat_list.append(number_stats_)
    df_stat_merged_=pd.concat(df_stat_list, axis=1)
    print(df_stat_merged_)
    df_stat_merged_.to_csv(f"{outdir}/{test_}/{test_}_number.csv")
```
