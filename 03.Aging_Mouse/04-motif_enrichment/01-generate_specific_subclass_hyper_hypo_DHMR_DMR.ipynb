```r
library(Seurat)
library(dplyr)
library(tidyr)
```

```r
# "indir" is a custom input path, and "outdir" is a custom output path.
# indir=""
# outdir=""
```

```r
aging_DHMR_state <- read.csv(paste0(indir,"/DHMR.subclass_diff_hyper_hypo_info.csv"))
```

```r
head(aging_DHMR_state);dim(aging_DHMR_state)
```

```r
unique(aging_DHMR_state$cluster)
```

```r
range(aging_DHMR_state$diff)
```

```r
aging_DHMR_state_filter <- aging_DHMR_state %>%
  mutate(
    start = as.numeric(gsub(".*_(\\d+)_.*", "\\1", chrom)),
    end = as.numeric(gsub(".*_(\\d+)$", "\\1", chrom)),
    length = end - start
  ) %>%
  filter(length >= 200) %>%
  select(-start, -end, -length)
```

```r
head(aging_DHMR_state_filter);dim(aging_DHMR_state_filter)
range(aging_DHMR_state_filter$diff)
```

```r
length(aging_DHMR_state_filter$chrom);length(unique(aging_DHMR_state_filter$chrom))
```

```r
aging_DMR_state <- read.csv(paste0(indir,"/DMR.subclass_diff_hyper_hypo_info.csv"))
```

```r
head(aging_DMR_state);dim(aging_DMR_state)
```

```r
range(aging_DMR_state$diff)
```

```r
aging_DMR_state_filter <- aging_DMR_state %>%
  mutate(
    start = as.numeric(gsub(".*_(\\d+)_.*", "\\1", chrom)),
    end = as.numeric(gsub(".*_(\\d+)$", "\\1", chrom)),
    length = end - start
  ) %>%
  filter(length >= 200) %>%
  select(-start, -end, -length)
```

```r
head(aging_DMR_state_filter);dim(aging_DMR_state_filter)
range(aging_DMR_state_filter$diff)
```

```r
length(aging_DMR_state_filter$chrom);length(unique(aging_DMR_state_filter$chrom))
```

```r
aging_DHMR_state_final <- aging_DHMR_state_filter %>%
  group_by(cluster) %>%
  summarize(
    Specific_Hyper_Region = list(unique(chrom[hyper_hypo == 1 & !(chrom %in% aging_DHMR_state$chrom[aging_DHMR_state$cluster != cluster & aging_DHMR_state$hyper_hypo == 1])])),
    Specific_Hyper_Count = length(Specific_Hyper_Region[[1]]),
    Specific_Hypo_Region = list(unique(chrom[hyper_hypo == 0 & !(chrom %in% aging_DHMR_state$chrom[aging_DHMR_state$cluster != cluster & aging_DHMR_state$hyper_hypo == 0])])),
    Specific_Hypo_Count = length(Specific_Hypo_Region[[1]])
  ) %>%
  rename(Subclass = cluster) %>%
  mutate(
    Specific_Hyper_Region = sapply(Specific_Hyper_Region, function(x) paste0("['", paste(x, collapse = "', '"), "']")),
    Specific_Hypo_Region = sapply(Specific_Hypo_Region, function(x) paste0("['", paste(x, collapse = "', '"), "']"))
  )
```

```r
write.csv(aging_DHMR_state_final, paste0(outdir,"/01-dhmr_specific_hyper_hypo_DHMR_include_region.csv"))
```

```r
aging_DMR_state_final <- aging_DMR_state_filter %>%
  group_by(cluster) %>%
  summarize(
    Specific_Hyper_Region = list(unique(chrom[hyper_hypo == 1 & !(chrom %in% aging_DMR_state$chrom[aging_DMR_state$cluster != cluster & aging_DMR_state$hyper_hypo == 1])])),
    Specific_Hyper_Count = length(Specific_Hyper_Region[[1]]),
    Specific_Hypo_Region = list(unique(chrom[hyper_hypo == 0 & !(chrom %in% aging_DMR_state$chrom[aging_DMR_state$cluster != cluster & aging_DMR_state$hyper_hypo == 0])])),
    Specific_Hypo_Count = length(Specific_Hypo_Region[[1]])
  ) %>%
  rename(Subclass = cluster) %>%
  mutate(
    Specific_Hyper_Region = sapply(Specific_Hyper_Region, function(x) paste0("['", paste(x, collapse = "', '"), "']")),
    Specific_Hypo_Region = sapply(Specific_Hypo_Region, function(x) paste0("['", paste(x, collapse = "', '"), "']"))
  )
```

```r
write.csv(aging_DMR_state_final, paste0(outdir,"/01-dmr_specific_hyper_hypo_DMR_include_region.csv"))
```
